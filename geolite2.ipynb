```python
import geoip2.database
import urllib.request
import json
import pandas as pd
import itertools

def get_public_ip():
    try:
        with urllib.request.urlopen("https://api64.ipify.org?format=json") as response:
            data = json.load(response)
            return data["ip"]
    except Exception as e:
        return str(e)

def get_location_from_multiple_mmdb(ip_address, city_db, country_db, asn_db):
    try:
        result = {"ip": ip_address}
        
        # City-level details
        with geoip2.database.Reader(city_db) as city_reader:
            city_response = city_reader.city(ip_address)
            result["city"] = city_response.city.name or "Unknown"
            result["region"] = city_response.subdivisions.most_specific.name or "Unknown"
            result["country"] = city_response.country.name or "Unknown"
            result["latitude"] = city_response.location.latitude or "Unknown"
            result["longitude"] = city_response.location.longitude or "Unknown"
        
        # Country-level details
        with geoip2.database.Reader(country_db) as country_reader:
            country_response = country_reader.country(ip_address)
            result["country_iso_code"] = country_response.country.iso_code or "Unknown"
        
        # ASN details
        with geoip2.database.Reader(asn_db) as asn_reader:
            asn_response = asn_reader.asn(ip_address)
            result["asn"] = asn_response.autonomous_system_number or "Unknown"
            result["asn_org"] = asn_response.autonomous_system_organization or "Unknown"
        
        return result
    except geoip2.errors.AddressNotFoundError:
        return None
    except Exception as e:
        return {"error": str(e)}

# Generate a subset of IP addresses for testing
def generate_ip_addresses():
    # Example: Generate IP addresses in the range 1.1.1.0 to 255.255.255.0
    ip_range = itertools.product(range(1, 256), range(1, 256), range(0,1), range(0, 1))
    ip_addresses = ["{}.{}.{}.{}".format(*ip) for ip in ip_range]
    return ip_addresses

# List of IP addresses to check
ip_addresses = generate_ip_addresses()

# Paths to GeoLite2 databases
city_db = "GeoLite2-City.mmdb"
country_db = "GeoLite2-Country.mmdb"
asn_db = "GeoLite2-ASN.mmdb"

# List to store results
results = []

# Check each IP address
for ip in ip_addresses:
    data = get_location_from_multiple_mmdb(ip, city_db, country_db, asn_db)
    if data:
        results.append(data)
        print(f"Processed IP: {ip}")

# Create a DataFrame
df = pd.DataFrame(results)
```

```python
df.to_csv('ip_addresses_with_data_new.csv', index=False)

print("IP addresses with data have been saved to 'ip_addresses_with_data_new.csv'")
```
